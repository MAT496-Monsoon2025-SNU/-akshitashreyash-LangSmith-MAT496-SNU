```python
import os
from dotenv import find_dotenv, load_dotenv
from langsmith import Client
from pathlib import Path
load_dotenv(find_dotenv(), override=True)
client = Client()
examples = [
    ("What is RAG?", "RAG is Retrieval-Augmented Generation, which mixes external context with an LLM."),
    ("What is Perplexity AI?", "Perplexity AI is an AI search engine using real-time retrieval and LLMs."),
    ("How do I authenticate Perplexity API?", "Use your PPLX_API_KEY as a Bearer token in the Authorization header."),
]
dataset_id = ""
if not dataset_id:
    ds = client.create_dataset(dataset_name="Perplexity-RAG-Examples")
    if isinstance(ds, dict):
        dataset_id = ds["id"]
    else:
        dataset_id = ds.id
    print("Created dataset:", dataset_id)
else:
    print("Using existing dataset:", dataset_id)

inputs = [{"question": q} for q, _ in examples]
outputs = [{"answer": a} for _, a in examples]
client.create_examples(
    inputs=inputs,
    outputs=outputs,
    dataset_id=dataset_id,
)
print("Uploaded examples successfully!")
```

```python
def simple_retrieve(question: str):
    return [f"Retrieved snippet about: {question}"]
def simple_generate(question: str, documents):
    ctx = "\n".join(documents)
    return f"Answer: Based on context [{ctx}], here is a concise reply to your question."
def langsmith_rag(question: str):
    docs = simple_retrieve(question)
    answer = simple_generate(question, docs)
    return answer
```

```python
question = "What is RAG?"
langsmith_rag(question)
```

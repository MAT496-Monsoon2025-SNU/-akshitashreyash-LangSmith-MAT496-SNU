```python
import requests, os
from dotenv import load_dotenv
load_dotenv()
api_key = os.getenv("PERPLEXITY_API_KEY")
if not api_key:
    raise Exception("PERPLEXITY_API_KEY missing")

url = "https://api.perplexity.ai/chat/completions"
payload = {
    "model": "sonar",   # try "sonar", or "sonar-pro" if you have it
    "messages": [
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Say hello in one short sentence."}
    ],
    "max_tokens": 30
}
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

resp = requests.post(url, headers=headers, json=payload, timeout=30)

if resp.status_code == 200:
    print("🤖 Model replied:", resp.json()["choices"][0]["message"]["content"].strip())
else:
    print("❌ Error", resp.status_code, resp.text)
```

```python
# G. Minimal agent function that calls Perplexity
import requests, os
from dotenv import load_dotenv
load_dotenv()
API_KEY = os.getenv("PERPLEXITY_API_KEY")
URL = "https://api.perplexity.ai/chat/completions"

def ask_perplexity(prompt, model="sonar", max_tokens=80):
    payload = {
        "model": model,
        "messages": [
            {"role": "system", "content": "You are a helpful assistant."},
            {"role": "user", "content": prompt}
        ],
        "max_tokens": max_tokens
    }
    headers = {"Authorization": f"Bearer {API_KEY}", "Content-Type": "application/json"}
    r = requests.post(URL, headers=headers, json=payload, timeout=30)
    if r.status_code == 200:
        return r.json()["choices"][0]["message"]["content"].strip()
    else:
        return f"ERROR {r.status_code}: {r.text}"

# Example usage:
print(ask_perplexity("Give one short interesting fact about economics."))
```
